```python
# Sorting Student Scores using Bubble Sort

scores = [78, 45, 90, 67, 88, 54, 72, 99, 61, 83]

print("Original Scores:")
print(scores)

# Bubble Sort
n = len(scores)

for i in range(n):
    for j in range(0, n - i - 1):
        if scores[j] > scores[j + 1]:
            scores[j], scores[j + 1] = scores[j + 1], scores[j]

print("\nSorted Scores (Ascending Order):")
print(scores)
```
